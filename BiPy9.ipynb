FASTA to DataFrame

```python
import pandas as pd
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

# Let's quickly write a dummy FASTA file to work with
dummy_fasta = """>seq_1
ATGCGTAACGTGTAA
>seq_2
GCGCGCGCGCGCGCG
>seq_3
ATATATATATATATA
>seq_4
ATGCGCGCATATATA"""

with open("sample_genes.fasta", "w") as f:
    f.write(dummy_fasta)
    
print("Sample FASTA created!")
```

```python
# 1. Create an empty list to hold our data
data_list = []

# 2. Iterate through the FASTA file
for record in SeqIO.parse("sample_genes.fasta", "fasta"):
    
    # Extract the features we care about
    seq_id = record.id
    seq_len = len(record.seq)
    
    # Calculate GC content and convert to percentage
    gc_cont = gc_fraction(record.seq) * 100 
    
    # 3. Create a dictionary for this specific sequence
    seq_features = {
        "ID": seq_id,
        "Length": seq_len,
        "GC_Percent": round(gc_cont, 2)
    }
    
    # 4. Add it to our master list
    data_list.append(seq_features)

print(f"Extracted features for {len(data_list)} sequences.")
```

```python
# 5. The Magic Bridge: Convert list to DataFrame
df = pd.DataFrame(data_list)

# Let's look at our beautiful new table!
display(df)
```

```python
# Calculate mean GC content
print(f"Average GC: {df['GC_Percent'].mean()}%")
```

```python
# Filter for high GC sequences
high_gc_genes = df[df['GC_Percent'] > 45.0]
display(high_gc_genes)
```
