## Building a DNA Dot Plot from Scratch (NumPy & Matplotlib)

```python
import numpy as np
import matplotlib.pyplot as plt

# Define our two sequences
seq1 = "ATGCGTACGTAGCTAGCTAG"
# seq2 has a repeat of 'CGTA' inserted in the middle!
seq2 = "ATGCGTACGTACGTAGCTAG" 

print(f"Sequence 1 Length: {len(seq1)}")
print(f"Sequence 2 Length: {len(seq2)}")
```

```python
# 1. Create an empty matrix of zeros
# The size is (length of seq2, length of seq1)
dot_matrix = np.zeros((len(seq2), len(seq1)))

# 2. Fill the matrix
for i in range(len(seq2)):          # Rows (Y-axis)
    for j in range(len(seq1)):      # Columns (X-axis)
        
        # If the nucleotide in seq2 matches the one in seq1...
        if seq2[i] == seq1[j]:
            dot_matrix[i, j] = 1    # Place a dot (1)

print("Matrix built successfully!")
```

```python
# 3. Create the plot
plt.figure(figsize=(8, 8))

# imshow displays our 1s and 0s as a heatmap grid
# cmap='Greys' makes the 0s white and 1s black
plt.imshow(dot_matrix, cmap='Greys')

# 4. Add the DNA letters to the axes so we know what we are looking at!
plt.xticks(np.arange(len(seq1)), list(seq1))
plt.yticks(np.arange(len(seq2)), list(seq2))

# 5. Add Titles and Labels
plt.title("DNA Sequence Dot Plot", fontsize=16, fontweight='bold')
plt.xlabel("Sequence 1", fontsize=12)
plt.ylabel("Sequence 2", fontsize=12)

# Show the final plot
plt.tight_layout()
plt.show()
```
