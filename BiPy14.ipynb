```python
from Bio import Entrez
from Bio import SeqIO

# 1. ALWAYS provide your email! 
# NCBI will block you if you spam them anonymously.
Entrez.email = "your.email@example.com"
```

```python
# 2. Search the database
print("Searching NCBI for the rice EPSPS gene...")

# db = the database to search (nucleotide, protein, pubmed, etc.)
# term = your search query, just like the website
# retmax = maximum number of results to bring back
search_handle = Entrez.esearch(db="nucleotide", term="Oryza sativa EPSPS", retmax=1)

# Read the results into a Python dictionary
search_results = Entrez.read(search_handle)
search_handle.close()

# 3. Extract the Sequence ID
seq_id = search_results["IdList"][0]
print(f"Found Sequence ID: {seq_id}")
```

```python
# 4. Fetch the actual DNA sequence using the ID
print("Fetching FASTA data directly into Python...")

fetch_handle = Entrez.efetch(db="nucleotide", id=seq_id, rettype="fasta", retmode="text")

# Read the data using SeqIO, just like a local file!
record = SeqIO.read(fetch_handle, "fasta")
fetch_handle.close()

print("\n--- DOWNLOAD COMPLETE ---")
print(f"Header: {record.description}")
print(f"Length: {len(record.seq)} base pairs")
print(f"First 50 bases: {record.seq[:50]}...")
```

```python
# 5. (Optional) Save it locally for future use
with open("rice_epsps_gene.fasta", "w") as out_file:
    SeqIO.write(record, out_file, "fasta")
    
print("Saved to rice_epsps_gene.fasta")
```
