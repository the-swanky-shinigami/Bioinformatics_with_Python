```python
import pandas as pd
import random
```

```python
# Generate a dummy clinical TSV file
print("Generating clinical_data.tsv...")

sample_ids = [f"PATIENT_{i:04d}" for i in range(1, 1001)]
tissue_types = random.choices(['Tumor', 'Normal'], weights=[0.7, 0.3], k=1000)
expression_levels = [round(random.uniform(5.0, 150.0), 2) for _ in range(1000)]
mutation_status = random.choices(['Positive', 'Negative'], k=1000)

# Write to a TSV file manually to simulate real-world data
with open("clinical_data.tsv", "w") as f:
    f.write("Sample_ID\tTissue_Type\tGene_X_Expression\tBRCA_Mutation\n") # Header
    for i in range(1000):
        f.write(f"{sample_ids[i]}\t{tissue_types[i]}\t{expression_levels[i]}\t{mutation_status[i]}\n")

print("Done! We have our dataset.")
```

```python
# Load the TSV file into a Pandas DataFrame
# We use sep='\t' because it's a Tab-Separated file
df = pd.read_csv("clinical_data.tsv", sep='\t')

# .head() shows us just the first 5 rows so we don't spam our screen
display(df.head())
```

```python
# Count the values in a specific column
tissue_counts = df['Tissue_Type'].value_counts()
print("Tissue Types Breakdown:")
print(tissue_counts)
```

```python
# Filtering the DataFrame with multiple conditions
# We use & for 'AND'. Make sure to put conditions in parentheses!
high_risk_patients = df[(df['Tissue_Type'] == 'Tumor') & (df['BRCA_Mutation'] == 'Positive')]

print(f"\nFound {len(high_risk_patients)} high-risk patients.")
display(high_risk_patients.head())
```

```python
# Calculate the mean of a column
avg_expression = high_risk_patients['Gene_X_Expression'].mean()

print(f"Average Gene X expression in high-risk patients: {avg_expression:.2f}")

# Save our filtered results to a new CSV to share with the lab
high_risk_patients.to_csv("high_risk_filtered.csv", index=False)
print("Saved filtered data to high_risk_filtered.csv!")
```
