## Reading DNA from Files

#### Reading and Parsing the FASTA File

```python
# Open and read the FASTA file
with open('my_sequence.fasta','r') as f:
    fasta_content = f.readlines()

print(fasta_content)
```

```python
# Initialize variables to store the header and sequence
fasta_header = ""
fasta_sequence = ""

with open('my_sequence.fasta','r') as f:
    for line in f:
        if line.startswith('>'):
            fasta_header = line.strip()
        else:
            fasta_sequence += line.strip()

print(f"Header found {fasta_header}")
print(f"Sequence found: {fasta_sequence}")
```

#### Using Our Parsed Sequence

```python
# Now we can analyze the sequence we loaded from the file
seq_length = len(fasta_sequence)
g_count = fasta_sequence.count('G')
c_count = fasta_sequence.count('C')

gc_content = ((g_count + c_count) / seq_length) * 100

print(f"The sequence length is {seq_length} bp")
print(f"The GC content is {gc_content:.2f}%")
```

### Pasing Multi-FASTA Files

```python
# Initialize an empty dictionary to hold our sequences
sequences = {}
current_sequence_id = None # To keep track of the current sequence

# Open and parse the multi-FASTA file
with open('genes.fasta', 'r') as f:
    for line in f:
        line = line.strip() # Remove leading/trailing whitespace
        if line.startswith('>'):
            # This is a header line. It's the key for our dictionary.
            current_sequence_id = line[1:] # Store the ID (without the '>')
            sequences[current_sequence_id] = "" # Initialize the sequence string
        else:
            # This is a sequence line. Append it to the current sequence.
            if current_sequence_id:
                sequences[current_sequence_id] += line

# Let's see what we got!
print(sequences)
```

```python
# Access a specific sequence and calculate its length
mouse_gene_id = 'gene_B_mouse P53 tumor suppressor'
mouse_gene_seq = sequences[mouse_gene_id]
print(f"Length of the mouse gene is: {len(mouse_gene_seq)} bp")
```

```python
# Loop through the dictionary and analyze each sequence
for seq_id, sequence in sequences.items():
    seq_length = len(sequence)
    g_count = sequence.count('G')
    c_count = sequence.count('C')
    gc_content = ((g_count + c_count) / seq_length) * 100
    
    # We can split the ID to get just the gene name for cleaner printing
    gene_name = seq_id.split(' ')[0]
    print(f"{gene_name}: GC Content is {gc_content:.2f}%")
```
