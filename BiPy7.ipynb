```python
import mygene
import pandas as pd
```

```python
# Initialize the MyGene tool
mg = mygene.MyGeneInfo()
```

```python
# A list of mysterious Ensembl IDs
ensembl_ids = [
    "ENSG00000139618", # We'll find out what this is!
    "ENSG00000141510",
    "ENSG00000012048",
    "ENSG00000146648"
]
```

```python
#Perform the translation
print("Querying the MyGene database...")

results = mg.querymany(
    ensembl_ids,
    scopes = 'ensembl.gene', # Our input format
    fields = 'symbol, name, entrezgene', # What we want back
    species = 'human'
)

print(f"Returned {len(results)} results!")
```

```python
# Convert the messy dictionary into a clean Pandas DataFrame
df = pd.DataFrame(results)

# Drop some internal APi columns we don't need ('_id' and '_score')
df = df.drop(columns=['_id', '_score'])

# Display the clean table
display(df)
```
