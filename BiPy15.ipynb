```python
# Our two sequences (must be the same length for this to work perfectly)
seq_wildtype = "ATGCGTACGTAGCTAG"
seq_mutant   = "ATGCGGACGTAGCGAG"
```

```python
# THE BEGINNER WAY
print("--- Beginner Method ---")
for i in range(len(seq_wildtype)):
    if seq_wildtype[i] != seq_mutant[i]:
        print(f"Mutation at index {i}: {seq_wildtype[i]} changed to {seq_mutant[i]}")
```

```python
# THE PRO WAY
print("\n--- Pro Method (Zip) ---")
for i, (base1, base2) in enumerate(zip(seq_wildtype, seq_mutant)):
    if base1 != base2:
        print(f"Mutation at index {i}: {base1} -> {base2}")
```

```python
# THE ULTIMATE 1-LINER
mutations = [(i, b1, b2) for i, (b1, b2) in enumerate(zip(seq_wildtype, seq_mutant)) if b1 != b2]

print("\n--- 1-Liner Result ---")
print(mutations)
```
