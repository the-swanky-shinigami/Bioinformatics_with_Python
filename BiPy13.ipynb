### Fast Counting

Short on time? In today's quick Bioinformatics Pro-Tip, we are upgrading our Python code! When you first learn Python, you use .count() to find nucleotide frequencies. But when your sequences get massive, that method is incredibly slow. Today, I'll show you how to use Python's built-in collections.Counter module to count DNA bases (or anything else) instantly and efficiently in just one line of code.

```python
import random

# Generate a random DNA sequence of 1 million bases
bases = ['A', 'T', 'C', 'G']
long_dna = ''.join(random.choices(bases, k=1000000))
print(f"Generated {len(long_dna)} base pairs.")
```

```python
# THE BEGINNER WAY
count_A = long_dna.count('A')
count_T = long_dna.count('T')
count_C = long_dna.count('C')
count_G = long_dna.count('G')

print(f"A: {count_A}, T: {count_T}, C: {count_C}, G: {count_G}")
```

```python
from collections import Counter

# THE PRO WAY
# Pass the entire string into Counter just once
base_counts = Counter(long_dna)

print(base_counts)
```

```python
# Find the single most common nucleotide
most_abundant = base_counts.most_common(1)

print(f"The most common base is: {most_abundant}")
```
