# The Venn Diagram Code

```python
import time

# Two lists of gene identifiers
breast_cancer_genes = ["BRCA1", "TP53", "EGFR", "PTEN", "MYC", "PIK3CA"]
lung_cancer_genes   = ["KRAS", "EGFR", "ALK", "TP53", "ROS1", "BRAF"]
```

```python
# THE BEGINNER WAY
common_genes = []

for gene in breast_cancer_genes:
    if gene in lung_cancer_genes:
        common_genes.append(gene)

print(f"Beginner Method found: {common_genes}")
```

Sets

```python
# THE PRO WAY (Intersection)
set1 = set(breast_cancer_genes)
set2 = set(lung_cancer_genes)

# Find genes in BOTH lists
pro_common = set1 & set2 

print(f"Pro Method found: {pro_common}")
```

```python
# Difference: Genes in set1 but NOT in set2
unique_to_breast = set1 - set2
print(f"Unique to Breast Cancer: {unique_to_breast}")
```

```python
# Union: Combine both, keeping only unique values
all_unique_genes = set1 | set2
print(f"All unique genes involved: {all_unique_genes}")
```

```python
# Symmetric Difference: Genes in either set, but NOT in both
exclusive_genes = set1 ^ set2
print(f"Mutated in one, but not both: {exclusive_genes}")
```
