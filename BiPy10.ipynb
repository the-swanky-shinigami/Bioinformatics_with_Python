Visualization

```python
# Install seaborn if you haven't already
#!pip install seaborn pandas matplotlib

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

print("Generating dummy genomic dataset...")

# Simulating 500 Traditional and 500 Improved sequences
categories = ['Traditional'] * 500 + ['Improved'] * 500

# Traditional might have slightly lower GC and shorter lengths
gc_trad = np.random.normal(loc=42.0, scale=3.0, size=500)
len_trad = np.random.normal(loc=1200, scale=200, size=500)

# Improved might have slightly higher GC and longer lengths
gc_imp = np.random.normal(loc=46.0, scale=2.5, size=500)
len_imp = np.random.normal(loc=1500, scale=150, size=500)

# Combine into a Pandas DataFrame
df = pd.DataFrame({
    'Variety': categories,
    'GC_Content': np.concatenate([gc_trad, gc_imp]),
    'Sequence_Length': np.concatenate([len_trad, len_imp])
})

display(df.head())
```

```python
# Set the visual theme for publication quality
sns.set_theme(style="whitegrid", palette="Set2")

# Create the Boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Variety', y='GC_Content')

# Add titles and labels
plt.title("GC Content Comparison: Traditional vs. Improved Varieties", fontsize=14, fontweight='bold')
plt.xlabel("Crop Variety", fontsize=12)
plt.ylabel("GC Content (%)", fontsize=12)

# Save it in high resolution (300 dpi is standard for journals!)
plt.savefig("gc_boxplot.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Create a Scatterplot with a hue
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, 
    x='Sequence_Length', 
    y='GC_Content', 
    hue='Variety',  # This magically color-codes the dots!
    alpha=0.7       # Makes the dots slightly transparent
)

plt.title("Sequence Length vs. GC Content", fontsize=14, fontweight='bold')
plt.xlabel("Sequence Length (bp)", fontsize=12)
plt.ylabel("GC Content (%)", fontsize=12)

plt.savefig("length_vs_gc_scatter.png", dpi=300, bbox_inches='tight')
plt.show()
```
