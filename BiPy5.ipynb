### Finding DNA Motifs and Patterns

```python
sequence = "GCTAGGAATTCGCTAGCTAGGAATTCGCA"
motif = "GAATTC"

# Find the first occurrence
pos = sequence.find(motif)
print(f"Motif found at index: {pos}")
```

```python
# Find all occurrences of the motif
start = 0
while True:
    start = sequence.find(motif, start)
    if start == -1: break # No more found
    print(f"Match at: {start}")
    start += 1 # Move forward to keep searching
```

```python
import re

# DNA sequence with some variations
seq = "GCTATAAAAGCTATATAAAGCTAG"

# Pattern: TA, then A or T, then AA
# This covers TATAA and TAAAA
pattern = "TA[AT]AA"

for match in re.finditer(pattern, seq):
    print(f"Found {match.group()} at position {match.start()}")
```

```python
# Find ATG followed by any 3 bases and then a GGC
complex_pattern = "ATG...GGC"
my_dna = "ATGCCCGGCTAGCTAGATGAAAGGC"

for m in re.finditer(complex_pattern, my_dna):
    print(f"Complex motif found: {m.group()} at {m.start()}")
```
