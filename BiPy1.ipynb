## Bioinformatics with Python 

#### 1. Representing DNA in Python

```python
dna_seq = 'ATGAACGTTAGACC'
print(dna_seq)
```

#### 2. Basic Sequence Properties

```python
seq_length = len(dna_seq)
print(f"The length of the sequence is: {seq_length} bases")
```

```python
a_count = dna_seq.count('A')
t_count = dna_seq.count('T')
c_count = dna_seq.count('C')
g_count = dna_seq.count('G')

print(f"A's: {a_count}")
print(f"T's: {t_count}")
print(f"C's: {c_count}")
print(f"G's: {g_count}")
```

```python
gc_content = ((g_count + c_count) / seq_length) * 100
print(f"The GC content is: {gc_content:.2f}%")
```

#### 3. Transcription

```python
# Transcribe DNA to RNA (replace T with U)
rna_seq = dna_seq.replace('T', 'U')
print(f"Original DNA: {dna_seq}")
print(f"Transcribed RNA: {rna_seq}")
```

#### 4. The Reverse Compliment

```python
# Step 1: Find the complement
# A -> T, T -> A, C -> G, G -> C

complement_dict = {'A': 'T', 'T' : 'A', 'C' : 'G', 'G' : 'C'}
complement_seq = "".join([complement_dict[base] for base in dna_seq])

print(f"Original sequence: {dna_seq}")
print(f"Complement sequence: {complement_seq}")
```

```python
# Step 2: Reverse the complement to get the reverse complement
reverse_complement_seq = complement_seq[::-1]

print(f"Original sequence:              5'-{dna_seq}-3'")
print(f"Reverse Complement sequence:    3'-{reverse_complement_seq}-5'")
```
