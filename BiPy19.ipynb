```python
# A standard 20-base PCR primer
primer_seq = "ATGCGATCGATCGATCGTAC"

# Manual Wallace Rule calculation
# (A+T) * 2 + (C+G) * 4
count_A = primer_seq.count('A')
count_T = primer_seq.count('T')
count_C = primer_seq.count('C')
count_G = primer_seq.count('G')

wallace_tm = ((count_A + count_T) * 2) + ((count_C + count_G) * 4)

print(f"Primer Sequence: {primer_seq}")
print(f"Estimated Tm (Wallace): {wallace_tm} °C")
```

```python
from Bio.SeqUtils import MeltingTemp as mt
from Bio.Seq import Seq

# Convert our string to a BioPython Seq object
my_primer = Seq("ATGCGATCGATCGATCGTAC")

# Calculate Tm using the Wallace rule (just to check our math!)
bio_wallace = mt.Tm_Wallace(my_primer)
print(f"BioPython Wallace Tm: {bio_wallace} °C")
```

```python
# Calculate Tm using Nearest Neighbor Thermodynamics
precise_tm = mt.Tm_NN(my_primer)

print(f"Highly Precise NN Tm: {precise_tm:.2f} °C")
```

```python
# Simulating a custom PCR buffer mix
# 50 millimolar Sodium, 1.5 millimolar Magnesium
custom_tm = mt.Tm_NN(my_primer, Na=50, Mg=1.5)

print(f"Custom Buffer Tm: {custom_tm:.2f} °C")
```
